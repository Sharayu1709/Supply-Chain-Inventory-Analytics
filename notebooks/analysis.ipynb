```python
df["Sales"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["Cost"] = df["Quantity"] * df["UnitCost"]
```

```python
df["Profit"] = df["Sales"] - df["Cost"] - df["ShippingCost"]
```

```python
print("Profitable Orders:", (df["Profit"] > 0).sum())
print("Loss-Making Orders:", (df["Profit"] < 0).sum())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Profit Margin:", round((total_profit / total_sales) * 100, 2), "%")
```

```python
below_reorder = df[df["StockLevel"] < df["ReorderPoint"]]
below_reorder.shape
```

```python
at_or_below_reorder = df[df["StockLevel"] <= df["ReorderPoint"]]
at_or_below_reorder.shape
```

```python
df["ReorderStatus"] = df.apply(
    lambda row: "Reorder Needed"
    if row["StockLevel"] <= row["ReorderPoint"]
    else "Stock OK",
    axis=1
)

df["ReorderStatus"].value_counts()
```

```python
inventory_summary = df.groupby("Product").agg(
    Avg_Stock_Level=("StockLevel", "mean"),
    Avg_Reorder_Point=("ReorderPoint", "mean"),
    Reorder_Needed_Count=("ReorderStatus", lambda x: (x == "Reorder Needed").sum())
).sort_values("Reorder_Needed_Count", ascending=False)

inventory_summary.round(2)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../database/supply_chain.db")

print("Connected to database successfully!")
```

```python
pd.read_sql_query("""
SELECT COUNT(DISTINCT OrderID) AS Total_Orders
FROM supply_chain;
""", conn)
```

```python
pd.read_sql_query("""
SELECT ROUND(SUM(Sales), 2) AS Total_Sales
FROM supply_chain;
""", conn)
```

```python
pd.read_sql_query("""
SELECT ROUND(SUM(Profit), 2) AS Total_Profit
FROM supply_chain;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    Product,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM supply_chain
GROUP BY Product
ORDER BY Total_Sales DESC
LIMIT 5;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    Supplier,
    COUNT(*) AS Total_Orders,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    ROUND(AVG(LeadTimeDays), 2) AS Avg_Lead_Time
FROM supply_chain
GROUP BY Supplier
ORDER BY Total_Sales DESC;
""", conn)
```
